# <center>End-to-End Restaurant Analytics and Machine Learning Solutions</center>

### 🔷 Introduction

In today’s data-driven world, the food industry relies heavily on data to understand customer preferences, improve services, and make informed business decisions. This project focuses on applying machine learning and data analysis techniques to a real-world restaurant dataset to extract meaningful insights and build predictive models.

As part of the internship with Cognifyz Technologies, this project involved solving multiple tasks including prediction, classification, and geographical analysis, providing hands-on experience in real-world data science workflows.

### 🔷 About the Tasks

🔹 Task 1: Restaurant Rating Prediction
- Built regression models using Linear Regression (model) and Decision Tree Regressor (model1)
- The Decision Tree model significantly outperformed Linear Regression (0.90 vs 0.20 R² score), indicating strong non-linear relationships in the dataset
- Analysis showed that votes, price range, and service features (online delivery, table booking) have a major impact on ratings
- Graphs and model results highlighted that simple linear models are not suitable for this type of real-world data

👉 Conclusion: Restaurant ratings depend on multiple interacting factors, and tree-based models are more effective in capturing these complex patterns.

🔹 Task 3: Cuisine Classification
- Developed classification models to predict cuisine types using encoded features
- Model accuracy was relatively low due to multi-label nature of cuisines, class imbalance, and similarity between cuisine types
- Tables and evaluation metrics showed that frequent cuisines were predicted better, while rare cuisines were often misclassified
- Simplifying cuisines to a single label improved model feasibility but introduced some bias

👉 Conclusion: Cuisine classification is a challenging problem due to overlapping categories and imbalanced data, requiring careful preprocessing and feature handling.

🔹 Task 4: Location-Based Analysis
- Visualized restaurant distribution using latitude and longitude
- Graphs showed that restaurants are highly concentrated in major cities
- Tables revealed variation in average ratings and price range across locations
- Analysis indicated that higher restaurant density does not always correspond to higher ratings
- Identified that restaurants cluster in commercial and high-demand areas

👉 Conclusion: Location plays a significant role in restaurant distribution, pricing, and performance, with clear differences observed across cities.

### 🔷 Key Insights and Findings

🔹 Model Performance

- Decision Tree Regressor significantly outperformed Linear Regression, highlighting the presence of non-linear relationships in the data

🔹 Customer Behavior

- Features like votes, pricing, and service availability strongly influence restaurant ratings

🔹 Cuisine Trends

- Popular cuisines dominate predictions, while rare cuisines are harder to classify due to class imbalance

🔹 Location Patterns

- Restaurants are concentrated in major cities, indicating higher demand in urban areas
- Higher restaurant density does not always guarantee higher ratings

🔹 Geographical Influence

- Price range, cuisine preference, and ratings vary significantly by location
- Restaurants tend to cluster in high-traffic and commercial areas

🔹 Data Challenges

- Handling multi-label cuisine data, missing values, and high-cardinality features was crucial for improving model performance
